# ``Series`` Indexing and Selection

We introduce more about ``Series`` indexing and selection in this notebook.

A ``Series`` builds on the dictionary-like interface and provides array-style item selection via the same basic mechanisms as NumPy arrays – that is, *slices*, *masking*, and *fancy indexing*.

Examples of these are as follows:

```python
import pandas as pd
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data
```

```python
data['b']
```

```python
'a' in data
```

```python
data['e'] = 1.25
data
```

```python
data['a'] = 5.25
data
```

```python
# slicing by the explicit indices
# note the inclusion of the 'stop' index
data['a':'c']
```

```python
# slicing by the implicit integer indices
# note the exclusion of the 'stop' index (as that in NumPy)
data[0:2]
```

```python
# masking
#(data > 0.3) & (data < 0.8)
data[(data > 0.3) & (data < 0.8)]
```

์<u>Note</u>: 
Among examples discussed above, slicing may be the source of the most confusion.
Notice that when slicing with an explicit index (i.e., ``data['a':'c']``), the final index is *included* in the slice, while when slicing with an implicit index (i.e., ``data[0:2]``), the final index is *excluded* from the slice.

## Indexers: ``loc`` and ``iloc``

Pandas provides some special *indexer attributes* that explicitly expose certain indexing schemes. 
These are not functional methods, but *attributes* that expose a particular slicing interface to the data in the Series.

---
First, the ``loc`` attribute allows indexing and slicing that <b>always references the explicit indices</b>:

```python
data
```

```python
data.index
```

```python
#note: since we are using the explicit indexing scheme, 
#      the inclusion of the 'stop' index is applied 
data.loc['b':'d']
```

---
Second, the ``iloc`` attribute allows indexing and slicing that **always references the implicit Python-style index**:

```python
data.iloc[1]
```

```python
#note: since we are using the implicit indexing scheme, 
#      the exclusion of the 'stop' index is applied 
data.iloc[1:3]
```

# ``DataFrame`` Indexing and Selection

## DataFrame As a Collection of Series

(มองว่า DataFrame is made up from a number of Series.)

### Column Access: access/select an entire column
Every column of the DataFrame (i.e. a Series) can be accessed using *dictionary-style indexing* via the column name.

Equivalently, we can use *attribute-style access* to select an entire column when column names are strings.

```python
import pandas as pd

state_dict = {'population': [38332521, 26448193, 19651127, 19552860, 12882135],
              'area':       [423967,   695662,   141297,   170312,   149995],
              'county':     [58,       254,      62,       67,       152]}
states = pd.DataFrame(state_dict, 
                      index=['California', 'Texas', 'New York', 'Florida', 'Illinois'])
states  
```

```python
# select the column 'population' of the dataframe
states['population']         # dictionary-style access
#states.population           # (alternative notation) attribute-style column access
```

```python
# select the column 'area' of the dataframe
#states['area']               # dictionary-style access
states.area                 # (alternative notation) attribute-style column access
```

<u>Remark</u>: 

Though the attribute-style access is a useful shorthand, keep in mind that it does not work for all cases! Specifically, if the column names are not strings, or if the column names conflict with methods of the DataFrame, this attribute-style access is not possible. 

For example, the DataFrame has a pop() method, so data.pop will point to this rather than the "pop" column

Adding a new column into a dataframe

The example below shows a preview of the straightforward syntax of element-by-element arithmetic between Series objects.

## DataFrame As a Two-Dimensional Arrays

we can also view the DataFrame as an enhanced *two-dimensional array*. This implies that we can specify both *row* and *column* indices to select the desired portion of data frame. 

### Applying the subscript operator ``[]`` directly to DataFrame

This manner works with limitations. Let's examine the following codes. 

```python
states
```

```python
# Select the first three rows 
states[:3]
```

```python
# Note! adding a new column called 'density' into the dataframe 'states'
states['density'] = states['population']/states['area']
states
```

```python
# Select 'area' and 'density' of all states in dataframe
states[['area','density']]
```

```python
# however, using the subscript operator, we cannot specify both rows and columns we want
states[:3, ['area','density']]
```

### Using the indexers ``loc`` and ``iloc``

The two indexers ``loc`` and ``iloc`` provide more powerful tools when selecting a desired portion inside a data frame.

```python
states
```

```python
states.loc['Florida','county'] = 50  
states
```

```python
# using the 'iloc' indexer to change the number of counties in Florida back to 67
# display the result
```

```python
# this 'iloc' indexer implies the use of implicit integer indexing 
states.iloc[:3, :2]                  
```

```python
# this 'loc' indexer implies the use of explicit indexing 
states.loc[:'Illinois', 'county':'density']
```

```python
# note that we slice two non-consecutive columns of the first two rows
states.loc[:'New York', ['area','density']]
```
