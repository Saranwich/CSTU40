## Pandas Series

### Pandas Series: Overview

A Pandas Series is a one-dimensional array of indexed data. It can be created from a list or array as follows:

```python
arr = np.array([0.25, 0.5, 0.75, 1.0])
arr
```

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0])
data
```

```python
type(data)
```

The ``Series`` wraps both a sequence of *values* and a sequence of *indices*, which we can access with the ``values`` and ``index`` attributes.

 The values are simply a familiar NumPy array:

```python
data.items()
```

```python
list(data.items())  
```

```python
data.values   
```

The index is an array-like object of type ``pd.Index``.

```python
data.index                          # equivalent to calling a function data.keys()
```

```python
type(data.index)
```

Like NumPy arrays, the Pandas series *data* can be accessed by the associated index via the familiar Python square-bracket notation:

```python
data[1:3]         # Note! the result indicates the exclusion of index 3
```

### Pandas Series: Fancy Indexing

The ``Series`` object is basically interchangeable with a one-dimensional NumPy array.
The essential difference is the presence of the index: while the Numpy Array has an *implicitly defined* integer index used to access the values, the Pandas ``Series`` has an *explicitly defined* index associated with the values.

This explicit index definition gives the ``Series`` object additional capabilities. For example, the index need not be an integer, but can consist of values of any desired type.

For example, if we wish, we can use strings as an index:

```python
# Example 1:
# Note that the default Integer indices can still be used.
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])   
data
```

![Pandas_Series_String_Index_1](https://drive.google.com/uc?export=view&id=16IZx43YKn-sY8yOxqMitNnqAsEBt1siY)

```python
data.values
```

```python
data.index  # equivalent to calling the function data.keys()
```

```python
data['a']
```

```python
# Example 2:
# Note that the default Integer indices can still be used.
months = pd.Series([31, 28, 31, 30, 31],
                   index=['jan', 'feb', 'mar', 'apr', 'may'])
```

![Pandas_Series_String_Index_1](https://drive.google.com/uc?export=view&id=1IA9MwXG9QgLR6T5lk6lZlMu_lbP_D7tz)

```python
months.index
```

```python
months
```

```python
months[4]
```

```python
months['may']
```

### Pandas Series: Constructing ``Series`` Objets from Dictionaries

The ``Series``-as-dictionary analogy can be made even more clear by constructing a ``Series`` object directly from a Python dictionary:

```python
# By default, a Series will be created 
# where the 'index' is drawn from the sorted dictionary keys
population_dict = {'California': 38332521,
                   'Texas': 26448193,
                   'New York': 19651127,
                   'Florida': 19552860,
                   'Illinois': 12882135}
population = pd.Series(population_dict) 
population
```

```python
population.index
```

```python
population['New York']
```

Unlike a dictionary, the Series also supports array-style operations such as slicing:

Please note the inclusion of 'Florida' in the slice.

```python
population[1:3]
```

```python
population['Texas':'Florida'] 
```

## Pandas DataFrame

***
A two dimensional labeled data structure with columns of potentially different types.
***

The next fundamental structure in Pandas is the ``DataFrame``.
If a ``Series`` is an analog of a *one-dimensional array* with flexible indices, a ``DataFrame`` is an analog of a *two-dimensional array* with both flexible row indices and flexible column names. 

Just as you might think of a two-dimensional array as an ordered sequence of aligned one-dimensional columns, you can think of a ``DataFrame`` as a sequence of aligned ``Series`` objects. Here, by "aligned" we mean that they share the same index.

![Pandas_Series_vs_DataFrame](https://drive.google.com/uc?export=view&id=1mLxj8kI7pzBnSCO12VpTrTHB4yJGN-Io)

# Retrieving Series/DataFrame Information

```python
import pandas as pd
# create a Series called 'data'
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])   
data
```

```python
# create a DataFrame called 'states'
state_dict = {'population': [38332521, 26448193, 19651127, 19552860, 12882135],
              'area':       [423967,   695662,   141297,   170312,   149995],
              'county':     [58,       254,      62,       67,       152]}
states = pd.DataFrame(state_dict, 
                      index=['California', 'Texas', 'New York', 'Florida', 'Illinois'])
states  
```
